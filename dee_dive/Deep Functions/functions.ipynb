```python
'''create a function which will print values until 10 using generators'''

def count_up_to(n):
    for i in range(1,n):
        yield i
    gen = count_up_to(n=10)

    print(next(gen)) 
    print(next(gen)) 
    print(next(gen)) 
```

```python
List_01=['1','2','3','4']
def my_Decorator(func):
    def wrapper(List_01):
        int_list = []
        for i in List_01:
            int_list.append(int(i))
        func(int_list)
    return wrapper    

@my_Decorator           
def originalFunc(List_01):
    newDict={}
    for i in List_01:
        newDict[i] = i**2
    print(newDict)  
originalFunc(List_01)
```
