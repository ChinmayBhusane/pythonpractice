```python
class student:
    def __init__(self):
        self.name = "rahul"  #public
        self._age = 20       #protected
        self.__marks =90     #private

    def display (self):
        print(self.name)    #public
        print(self._age)    #protected
        print(self.__marks) #private

obj = student()
obj.display()
```

```python
#outside theclass 
class student:
    def __init__(self):
        self.name = "rahul"  #public
        self._age = 20       #protected
        self.__marks =90     #private

obj = student()

#access outside the class


print(obj.name)    #public
print(obj._age)    #protected
print(obj.__marks) #private

```

```python
# Inside child class Inheritance
class ParentClass:             #parent class
    def __init__(self):
        self.name = "Rahul"    # Public
        self._age = 20         # Protected
        self.__marks = 90      # Private
class ChildClass(ParentClass): #child class
    def display(self):
        print(self.name)       # Public
        print(self._age)       # Protected
        print(self.__marks)    # Private

child_obj = ChildClass()
child_obj.display()
```

```python
class Student:
    def __init__(self):
        self.name = "Rahul"    # Public
        self._age = 20         # Protected
        self.__marks = 90      # Private


class Teacher:
    def display(self):
        student_Obj = Student() 
        print(student_Obj.name)
        print(student_Obj._age)
        print(student_Obj.__marks)

tech_Obj = Teacher()
tech_Obj.display()
```
