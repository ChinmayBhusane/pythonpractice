Method OVERloading



```python
#with overloading
class myClass:
    def sum_Overload(self,a= None,b= None,c= None,d=None):
        if a!=None and b!=None and c!=None and d!=None:
            return a+b+c+d
        elif a!=None and b!=None and c!=None:
            return a+b+c
        elif a!=None and b!=None:
            return a+b
        else:
            return ("Please provide at least two numbers to sum.")

my_Obj = myClass() 
my_Obj.sum_Overload(1,2,3,4)

```

```python
#simpler method without overloading
class myClass:
    def sum_4(self,a,b,c,d):
        return a+b+c+d
    
    def sum_3(self,a,b,c):
        return a+b+c
    
    def sum_2(self,a,b):
        return a+b

my_obj = myClass()
my_obj.sum_3(1,2,3)
my_obj.sum_2(1,1)
my_obj.sum_4(1,2,3,4)
```

```python
#simpler method without overloading
class myClass:
    def sum_4(self,a,b,c,d):
        return a+b+c+d
    
    def sum_3(self,a,b,c):
        return a+b+c
    
    def sum_2(self,a,b):
        return a+b

my_obj = myClass()
my_obj.sum_3(1,2,3)
my_obj.sum_2(1,1)
my_obj.sum_4(1,2,3,4)
```

#Method overriding

```python
class animal:
    def sound(self):
        print("Animal makes a sound")
class dog(animal):
    def sound(self):
        return "barks"
class cat(animal):
    def sound(self):
        return "meows"

dog_obj = dog()
dog_obj.sound()
```

```python
#use of super keyword
class P_C_Interest:
    def interest(self):
        print( 0.011)

class C_C_Home_Interest(P_C_Interest):
    def interest(self):
        super().interest()
        print (0.08)
    
Home_oBJ = C_C_Home_Interest()
Home_oBJ.interest()
```
