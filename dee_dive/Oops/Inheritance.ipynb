```python
class ParentClass:
    def Care_Health_Insurance(self):
        return "We provide Health Insurance"


class ChildClass(ParentClass):
    def Eye_Care(self):
        return "Lasik eye surgery is provided"


Child_Obj_01 = ChildClass()

print(Child_Obj_01.Care_Health_Insurance())
print(Child_Obj_01.Eye_Care())
```
