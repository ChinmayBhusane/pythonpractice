```python
#add
mySet = {"apple", "banana", "mango"}
mySet.add('orange')
```

```python
#update
mySet = {"apple", "banana", "mango"}
mySet.update(["orange", "grapes"])
mySet
```

```python
#remove
mySet = {"apple", "banana", "mango"}
mySet.remove("banana")
```

```python
#discard
mySet = {"apple", "banana", "mango"}
mySet.discard("banana")
```

```python
#pop
mySet = {"apple", "banana", "mango"}
mySet.pop("")
```

```python
#add items
mySet = {"apple", "banana", "mango"}
mySet.add("pineapple")
mySet
```

```python
mySet = {"apple", "banana", "mango"}

abc = ["pineapple","grapes"]
mySet.update(abc)
mySet
```

```python
#delete
abcSet = {"apple", "banana", "cherry", "orange"}
del abcSet
```

```python
#clear
mySet = {"apple", "banana", "mango"}
mySet.clear()
```

```python
#copy
mySet = {"apple", "banana", "mango"}
mySet.copy()
```

```python
#union of set
abcSet1 = {"cherry", "orange"}
abcSet2 = {"apple", "banana"}

abc = abcSet1.union(abcSet2)
abc
```
