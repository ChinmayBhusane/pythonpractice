```python
newlist = []
for i in range(15):
    if i <5:
       newlist.append(i*2)
    elif i <10:
        newlist.append(i/2)
    else:
        newlist.append(i+2)
newlist   
```

```python
[i*2 if i<5 else i/2 if i<10 else i+2 for i in range(15)]
```

```python
#converrt to dict comprehension
{i:i*2 if i<5 else i/2 if i<10 else i+2 for i in range(15)}
```

```python
'''apply filter'''
fruitsList = [1,2,3,4,5,6,7,8,]
def extractonlyevennumbers(i):
    return i%2==0

abc = filter(extractonlyevennumbers,fruitsList)
print(abc)
print(dir(abc))

for i in abc:
    print(i)
```
