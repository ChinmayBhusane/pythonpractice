```python
#use of yield keyword
def newfunc(a):
    for i in range(1,a):
        yield i
values_01=newfunc(10)
print(next(values_01))
print(next(values_01))
print(next(values_01))
```

```python
def newfunc_01(a):
    for i in range (1,a):
        return i
vlaus= newfunc_01(10)
print(vlaus)
print(vlaus)
print(vlaus)
```
