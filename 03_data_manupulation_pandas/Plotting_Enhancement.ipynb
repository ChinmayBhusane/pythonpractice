```python
import pandas as pd
df = pd.read_excel("E:\ethans\Data\Swarn_Airline.xlsx")
df
```

```python
order = df.groupby('Airline')['Price'].sum().sort_values().index
order
```

```python
df['Date_of_Journey'] = pd.to_datetime(df['Date_of_Journey'])
df['Date_of_Journey'].dt.month_name()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="dark", palette="pastel")
plt.figure(figsize=(10, 5))
Airline_sort=df.groupby('Airline')['Price'].sum().sort_values().index

ax=sns.barplot(x=df['Airline'],y=df['Price'],errorbar=None,estimator='sum',order=Airline_sort,alpha=0.9)
ax.bar_label(ax.containers[0],rotation=90)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="dark", palette="pastel")

plt.figure(figsize=(10, 5))

Airline_sort = df['Airline'].value_counts().sort_values().index

ax = sns.countplot(
    x='Airline',
    data=df,
    order=Airline_sort,
    alpha=0.9
)

ax.bar_label(ax.containers[0], rotation=90)

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

```

```python
sns.set_theme(style="whitegrid", palette="pastel")
plt.figure(figsize=(10, 5))
Airline_sort=df.groupby('Airline')['Price'].count().sort_values().index

sns.lineplot(x=df['Airline'],y=df['Price'],errorbar=None,estimator='mean',alpha=0.9,linestyle='--')



plt.xticks(rotation=90)
plt.tight_layout()
plt.show
```
